```python
# ============================================================
# EVALUATION 3 — CLAIM-LEVEL FAITHFULNESS
# ============================================================
#
# PURPOSE
# -------
# Evaluate whether each atomic clinical claim generated by the
# nine RAG configurations is supported by the patient's original
# source notes.
#
# IMPORTANT
# ---------
# This notebook:
# - DOES NOT regenerate the nine RAG summaries.
# - DOES NOT regenerate atomic claims.
# - DOES NOT use NLI.
# - DOES NOT evaluate temporal ordering.
#
# It ONLY performs claim-level factual faithfulness evaluation.
#
# FINAL EVALUATOR
# ---------------
# Provider: OpenAI
# Model: gpt-5.6-luna
# Temperature: model default
#
# EVIDENCE RETRIEVAL
# ------------------
# Each claim is semantically matched against the COMPLETE
# source-of-truth patient record using Gemini embeddings.
# Top-5 source notes are provided to the evaluator.
#
# FAITHFULNESS
# ------------
# Supported claims / total claims * 100
# ============================================================

import json
import os
import time

import numpy as np
import pandas as pd
from google import genai
from openai import OpenAI
from sklearn.metrics.pairwise import cosine_similarity


EVALUATOR_PROVIDER = "OpenAI"
EVALUATOR_MODEL = "gpt-5.6-luna"
EVIDENCE_EMBEDDING_MODEL = "gemini-embedding-001"
TOP_K_EVIDENCE = 5

print("======================================")
print("EVALUATION 3 — CLAIM FAITHFULNESS")
print("======================================")
print("Evaluator provider :", EVALUATOR_PROVIDER)
print("Evaluator model    :", EVALUATOR_MODEL)
print("Evidence embeddings:", EVIDENCE_EMBEDDING_MODEL)
print("Evidence Top-K     :", TOP_K_EVIDENCE)
print("======================================")
```

```python
# ============================================================
# API CLIENTS
#
# OpenAI:
#   Used ONLY for SUPPORTED / UNSUPPORTED claim verification.
#
# Gemini:
#   Used ONLY for semantic evidence retrieval.
# ============================================================

openai_client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY")
)

gemini_client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

print("OpenAI client class :", type(openai_client).__name__)
print("OpenAI client module:", type(openai_client).__module__)
print("Evaluator model     :", EVALUATOR_MODEL)

print("\nGemini client ready.")
print("Evidence model      :", EVIDENCE_EMBEDDING_MODEL)
```
